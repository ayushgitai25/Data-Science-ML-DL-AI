```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings

embeddings = GoogleGenerativeAIEmbeddings(
  model="models/gemini-embedding-001"
)
```

```python
text = "This is a tutorial on Google Gemini Embedding"

query_result = embeddings.embed_query(text= text)
```

```python
## Text Loaders
from langchain_community.document_loaders import TextLoader

loader = TextLoader('text.txt')
docs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
  chunk_size = 100,
  chunk_overlap = 50
)

# final_documents = text_splitter.create_documents(docs) ## useful for text file
final_documents = text_splitter.split_documents(docs)
final_documents
```

```python
## Vector DB = ChromaDB

from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(
  final_documents,  ## Pass your documents here
  embedding = embeddings ## Pass your embedding model here
)
```

```python
query = "The Taj Mahal is a white marble mausoleum"

retreived_results = db.similarity_search(query)
print(retreived_results) ## Get similar result
```
