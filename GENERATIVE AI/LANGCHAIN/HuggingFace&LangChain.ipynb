### Accessing HGF models with API : Serverless API, which is beneficial to people using pro-accounts or enterprise hub.

```python
from langchain_huggingface import HuggingFaceEndpoint
```

```python
repo_id = "google/flan-t5-large"
llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    temperature=0.7,
    max_new_tokens=256,
    provider="auto"
)
llm
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
  """
    Question: {question}
    Answer:  Answer things step by step.                                       
  """
)
prompt
```

```python
chain = prompt | llm

chain
```

```python
result = chain.invoke({"question": "Who won the cricket world cup 2011"})
print(result)
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tok = AutoTokenizer.from_pretrained("google/flan-t5-large")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large")

inputs = tok("Question: Who won the cricket world cup 2011\nAnswer:", return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=64)
print(tok.decode(outputs[0], skip_special_tokens=True))
```
