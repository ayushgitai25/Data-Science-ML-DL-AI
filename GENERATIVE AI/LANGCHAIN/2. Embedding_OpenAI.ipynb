```python
embeddings_1024 = OpenAIEmbeddings(
  model = 'text-embedding-3-large',
  dimensions= 1024
)
```

```python
text = "This is a tutorial on OpenAI Embedding"

query_result = embeddings_1024.embed_query(text= text)
len(query_result)
```

## 1. Load

```python
## Text Loaders
from langchain_community.document_loaders import TextLoader

loader = TextLoader('text.txt')
docs = loader.load()
```

## 2. Split

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
  chunk_size = 100,
  chunk_overlap = 50
)

# final_documents = text_splitter.create_documents(docs) ## useful for text file
final_documents = text_splitter.split_documents(docs)
final_documents
```

# 3. Convert into vectors & Store in a Vector DB

```python
## Vector DB = ChromaDB

from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(
  final_documents,  ## Pass your documents here
  embedding = embeddings_1024 ## Pass your embedding model here
)
```

# 4. Do a similarity search 

```python
query = "The Taj Mahal is a white marble mausoleum"

retreived_results = db.similarity_search(query)
print(retreived_results) ## Get similar result
```
