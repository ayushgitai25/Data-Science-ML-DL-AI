```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq(
  model= "openai/gpt-oss-120b" ## Open-Source model :  This model "openai/gpt-oss-120b" is an open-source model released by OpenAI.
)
llm
```

```python
from langchain.schema import (
  AIMessage, HumanMessage, SystemMessage
)
```

```python
english_txt = """\
Once upon a time, in a small village nestled between rolling hills and dense forests, there lived a young boy named Oliver. He loved to explore the woods, climbing trees, and discovering hidden streams that glittered in the sunlight.

Every morning, Oliver would rise with the sun, eager to see what new adventures awaited him. The villagers often warned him about the dangers of wandering too far, but his curiosity was stronger than their caution.

One day, as he followed the song of a distant bird, he stumbled upon a clearing he had never seen before. In the center stood a massive, ancient oak tree, its branches stretching high into the sky. At the base of the tree was a small, wooden door, carved with strange symbols.

Intrigued, Oliver gently pushed the door open and discovered a spiral staircase leading downward. With a deep breath, he descended into the hidden world beneath the village—a world filled with glowing mushrooms, sparkling rivers, and creatures he had only read about in books.

As days turned into weeks, Oliver explored this magical underground realm, learning its secrets and forming friendships with the fantastical beings who lived there. He realized that courage, kindness, and curiosity were his greatest tools.

Eventually, he returned to his village, bringing back stories of wonder and mystery. The villagers listened in awe, and though some still worried, they could not deny the joy and knowledge Oliver shared. From that day forward, he became the village storyteller, inspiring generations to come with his tales of adventure and discovery.
"""
```

```python
chat_message = [
    SystemMessage(content="You are an expert with expertise in summarizing texts."),
    HumanMessage(content=f"Please summarize the following text:\n\n{english_txt}")
]
```

```python
llm.get_num_tokens(english_txt)
```

```python
llm(chat_message)
```

## Using Prompt Template (with Latest LangChain Implementation):

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template(
  """
  Write a concise summary of the following text.
  Text : {text},
  Convert the summary into following language: {language}
  """ 
)
```

```python
chain = prompt | llm
chain
```

```python
chain.invoke(
  {"text": english_txt, "language": "French"}
)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt | llm | output_parser
chain
```

```python
print(chain.invoke(
  {"text": english_txt, "language": "French"}
))
```

```python
print(chain.invoke(
  {"text": english_txt, "language": "Hindi"}
))
```
