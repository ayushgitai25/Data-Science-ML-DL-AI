### Ollama supports embedding models, making it possible to build retreival augmented generation (RAG) applications that combine text prompts with existing documents or other data.

Ollama supports several models specifically designed for generating embeddings. These models are ideal for tasks like semantic search and Retrieval-Augmented Generation (RAG), where you need to convert text into numerical vectors that capture its meaning. Some of the most popular and high-performing embedding models available on Ollama include:

Key Ollama Embedding Models
nomic-embed-text: This is a very popular choice and is often considered a general-purpose, high-performing embedding model. It's known for its effectiveness on both short and long text tasks.

mxbai-embed-large: A state-of-the-art embedding model from mixedbread.ai, this model is highly effective for high-precision semantic search and is known for its strong performance on benchmarks like the Massive Text Embedding Benchmark (MTEB).

all-minilm: This model is a great option if you need a lightweight and fast-processing model. While it may have fewer dimensions than larger models, it's very efficient and a good starting point for many applications.

snowflake-arctic-embed: This model is part of a suite of models by Snowflake and is specifically optimized for tasks involving code and technical documentation.

embeddinggemma: A 300M parameter model from Google, which is part of the Gemma family of models.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings = OllamaEmbeddings(
  ## By default it uses llama2
  # model= 'nomic-embed-text'
  model = "embeddinggemma"
)
```

```python
embeddings
```

```python
embedded = embeddings.embed_documents(
  [
    "I am using gemma-3 locally on my system.",
    "I am using Ollama which can be used for many other models."
  ]
)
```

```python
embedded[0]
```

```python
len(embedded[0])
```

```python
embeddings.embed_query("What can be used to locally run many models?")
```
