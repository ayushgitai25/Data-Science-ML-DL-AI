```python
!pip install nltk
```

```python
corpus = """Hello guys! Let's talk about NLTK? NLTK is a leading platform for building Python programs to work with human language data. 
It provides easy-to-use interfaces to over 50 corpora and lexical resources such as WordNet, along with 
a suite of text processing libraries for classification, tokenization, stemming, tagging, parsing, and 
semantic reasoning, wrappers for industrial-strength NLP libraries, and an active discussion forum.
NLTK's.
"""
```

```python
print(corpus)
```

## Tokenization
### Paragraph --> Sentences

```python
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

# Tokenization
## Corpus --> words
## Sentence --> words

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
  print(word_tokenize(sentence))
```

## Treating punctuation seperatley : apostrophe s is now treated as a single word 

```python
from nltk.tokenize import wordpunct_tokenize


for sentence in documents:
  print(wordpunct_tokenize(sentence))

## Diffrence is that : ['NLTK', "'", 's', '.']    s is now seperated we see.
```

## TreebankWordTokenizer : all full stop are not treated as seperate word. The last full stop is treated seperately only.

```python
from nltk import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

## diffrence : 'data.',  'forum.', '.'] 
```

```python
tokenizer.tokenize(corpus)
```
