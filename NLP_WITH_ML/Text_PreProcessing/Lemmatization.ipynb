#### The WordNetLemmatizer is a tool within the Natural Language Toolkit (NLTK) library in Python that uses the WordNet lexical database to find the base or dictionary form of a word, known as its lemma. It's used in Natural Language Processing (NLP) for tasks like text normalization, information retrieval, and search result improvement. 

#### Lemmatization:
#### The process of reducing a word to its base or dictionary form (its lemma). For example, "running" would be lemmatized to "run", and "better" to "good"

#### WordNet:
#### A large lexical database of English that groups words into sets of synonyms (synsets) and provides semantic relationships between them. 

#### Part of Speech (POS) Tagging:
#### WordNetLemmatizer often requires the part of speech of a word to accurately determine the lemma. For example, "running" can be a verb or a noun, and the correct POS tag is needed for proper lemmatization. 

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS- Noun-n
verb-v
adjective-a
abverb-r
'''
lemmatizer.lemmatize("going", pos = 'v')
```

```python
words = ['eating', 'eats', 'eaten', 'writes', 'writing', 'programming', 'programs', 'finally', 'finalizing', 'history']
```

```python
for word in words:
  print(word + "-->" + lemmatizer.lemmatize(word, pos = 'v'))
```

```python
for word in words:
  print(word + "-->" + lemmatizer.lemmatize(word, pos = 'n'))
```

```python
for word in words:
  print(word + "-->" + lemmatizer.lemmatize(word, pos = 'a'))
```
