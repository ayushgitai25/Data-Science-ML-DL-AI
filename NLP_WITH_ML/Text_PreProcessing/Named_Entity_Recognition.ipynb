```python
sentence = "Barack Obama was born in Hawaii and became President of the United States."
```

```python
import nltk
words=  nltk.word_tokenize(sentence)
```

```python
tag_elements = nltk.pos_tag(words)
```

```python
# Make sure you've downloaded necessary resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
```

```python
named_entities = nltk.ne_chunk(tag_elements)
```

```python
print(named_entities)
```
