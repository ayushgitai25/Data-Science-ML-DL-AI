```python
# A Part-of-Speech (POS) tag is a label that indicates the grammatical role of a word in a sentence, such as noun, verb, adjective, etc. Here's a list of common POS tags: 
# Noun (NN): Represents a person, place, thing, or idea (e.g., "cat", "house", "freedom"). 
# Pronoun (PRP): A word that replaces a noun (e.g., "he", "she", "it"). 
# Verb (VB): Represents an action or state of being (e.g., "run", "is", "think"). 
# Adjective (JJ): Describes a noun or pronoun (e.g., "big", "happy", "blue"). 
# Adverb (RB): Modifies a verb, adjective, or another adverb (e.g., "quickly", "very", "extremely"). 
# Preposition (IN): Shows the relationship between a noun or pronoun and other words in a sentence (e.g., "on", "in", "at"). 
# Conjunction (CC): Connects words, phrases, or clauses (e.g., "and", "but", "or"). 
# Determiner (DT): Limits or specifies a noun (e.g., "the", "a", "this"). 
# Interjection (UH): An exclamatory word that expresses emotion (e.g., "Wow!", "Oh!", "Hey!"). 
# Numeral (CD): Represents a number (e.g., "one", "2", "third"). 
# Proper Noun (NNP): A specific person, place, or thing (e.g., "John", "London", "Google"). 
# Possessive Pronoun (PRP$): Indicates ownership (e.g., "my", "his", "her"). 
# Wh-determiner (WDT): A determiner that introduces a question (e.g., "which", "what"). 
# Wh-pronoun (WP): A pronoun that introduces a question (e.g., "who", "whom"). 
# Wh-adverb (WRB): An adverb that introduces a question (e.g., "where", "when"). 
# Particle (RP): A word that combines with a verb to form a phrasal verb (e.g., "up" in "look up"). 
# Symbol (SYM): Represents a symbol or special character (e.g., $, %, &). 
# Foreign word (FW): Words from other languages. 
# Existential there (EX): Used to introduce a statement (e.g., "there is"). 
# Modal (MD): A verb that expresses modality (e.g., "can", "will"). 
# Possessive ending (POS): Indicates possession (e.g., "'s" in "John's"). 
# Predeterminer (PDT): Occurs before a determiner (e.g., "all" in "all the books"). 
# List marker (LS): Used in lists (e.g., "1" in "1. Item"). 
# Verb, past tense (VBD): (e.g., "took", "went"). 
# Verb, gerund/present participle (VBG): (e.g., "taking", "running"). 
# Verb, past participle (VBN): (e.g., "taken", "gone"). 
# Verb, present tense (VBP): (e.g., "take", "go"). 
# Verb, 3rd person singular present (VBZ): (e.g., "takes", "goes"). 
```

```python
corpus = """
When Earth becomes uninhabitable in the future, a farmer and ex-NASA pilot, Joseph Cooper, is tasked to pilot a spacecraft, 
along with a team of researchers, to find a new planet for humans.
In the near future around the American Midwest, Cooper, an ex-science engineer and pilot, is tied to his farming land with his 
daughter Murph and son Tom. As devastating sandstorms ravage Earth's crops, the people of Earth realize their life here is coming 
to an end as food begins to run out. Eventually stumbling upon a N.A.S.A. base 6 hours from Cooper's home, he is asked to go on a 
daring mission with a few other scientists into a wormhole because of Cooper's scientific intellect and ability to pilot aircraft 
unlike the other crew members. In order to find a new home while Earth decays, Cooper must decide to either stay, or risk never 
seeing his children again in order to save the human race by finding another habitable planet.
In Earth's future, a global crop blight and second Dust Bowl are slowly rendering the planet uninhabitable. 
Professor Brand, a brilliant N.A.S.A. physicist, is working on plans to save mankind by transporting Earth's population 
to a new home via a womhole. But first, Brand must send former N.A.S.A. pilot Cooper and a team of researchers through 
the wormhole and across the galaxy to find out which of three planets could be mankind's new home.
"""
```

```python
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
```

```python
documents
```

```python
from nltk.corpus import stopwords
```

```python
import nltk
nltk.download('stopwords')
```

```python
## We will find the POS tag

for i in range(len(documents)):
  words = nltk.word_tokenize(documents[i])

  words = [word for word in words if word not in set(stopwords.words('english'))]

  ## documents[i] = " ".join(words) ## converting all the list of words back to sentences
  pos_tag = nltk.pos_tag(words)
  print(pos_tag)
```
