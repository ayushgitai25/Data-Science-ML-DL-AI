
About Dataset
Context
This is a small subset of dataset of Book reviews from Amazon Kindle Store category.

Content
5-core dataset of product reviews from Amazon Kindle Store category from May 1996 - July 2014. Contains total of 982619 entries. Each reviewer has at least 5 reviews and each product has at least 5 reviews in this dataset.
Columns

asin - ID of the product, like B000FA64PK
-helpful - helpfulness rating of the review - example: 2/3.
-overall - rating of the product.
-reviewText - text of the review (heading).
-reviewTime - time of the review (raw).
-reviewerID - ID of the reviewer, like A3SPTOKDG7WBLN
-reviewerName - name of the reviewer.
-summary - summary of the review (description).
-unixReviewTime - unix timestamp.
Which file to use?
There are two files one is preprocessed ready for sentiment analysis and other is unprocessed to you basically have to process the dataset and then perform sentiment analysis

```python
import pandas as pd

data = pd.read_csv('all_kindle_review .csv')
data.head()
```

```python
## Positive review => 1, Negative Review => 0
data['rating'] = data['rating'].apply(
  lambda x : 0 if x < 3 else 1
)
data.head()
```

```python
df = pd.DataFrame(data)
```

```python
df.head()
```

```python
df['rating'].value_counts()
```

```python
## 1. Pre-Processing and Cleaning (removing stopwords, stemming/lemmatization, lower-casing, removing non alpha chars, etc.)
df['reviewText'] = df['reviewText'].str.lower()
df
```

```python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from bs4 import BeautifulSoup

df['reviewText'] = df['reviewText'].apply(
  lambda x : re.sub('[^a-z A-Z 0-9]+', '', x) ## remove special characters
)
df['reviewText'] = df['reviewText'].apply(
  lambda x : " ".join([y for y in x.split() if y not in stopwords.words('english')]) ## remove stopwords
)
df['reviewText'] = df['reviewText'].apply(  ## handle urls
  lambda x : re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '', str(x))
)
df['reviewText'] = df['reviewText'].apply(  ## remove html tags
    lambda x: BeautifulSoup(str(x), 'html.parser').get_text()
)
df['reviewText'] = df['reviewText'].apply( ## remove additional spaces
    lambda x: " ".join(x.split())
)
```

```python
df['reviewText']
```

```python
df
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

def lemmatize_words(text):
  return " ".join([lemmatizer.lemmatize(word) for word in text.split()]) 
```

```python
df['reviewText'] = df['reviewText'].apply(
  lambda x : lemmatize_words(x)
)
```

```python
df
```

```python
## 2. Train-Test Split
from sklearn.model_selection import train_test_split

X = df['reviewText']
y = df['rating']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.20)
```

```python
X_train
```

```python
X_test
```

```python
## 3. Vector Embeddings using : BOW, TF-IDF, Word2Vec
## 4. Train using ML algorithms

## Create BOG
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
## Here we go with non-binary that means count of words (ex. [2, 1, 0] count is considered here) will be considered
cv = CountVectorizer()
```

```python
X_train_bow = cv.fit_transform(X_train).toarray()
X_test_bow = cv.transform(X_test).toarray() ## to avoid data leakage
```

```python
X_train_bow
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tv = TfidfVectorizer()
```

```python
X_train_tfidf = tv.fit_transform(X_train).toarray()
X_test_tfidf = tv.transform(X_test).toarray() ## to avoid data leakage
```

```python
X_train_tfidf
```

```python
from sklearn.naive_bayes import GaussianNB

nb_model_bow = GaussianNB().fit(X_train_bow, y_train)
```

```python
nb_model_tfidf = GaussianNB().fit(X_train_tfidf, y_train)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
y_pred_bow = nb_model_bow.predict(X_test_bow)
```

```python
y_pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)
```

```python
print(f"BOW Accuracy : {accuracy_score(y_test, y_pred_bow)}")
```

```python
print(f"TF-IDF Accuracy : {accuracy_score(y_test, y_pred_tfidf)}")
```

```python
confusion_matrix(y_test, y_pred_bow)
```

```python
confusion_matrix(y_test, y_pred_tfidf)
```

```python
import gensim
from gensim.models import Word2Vec, KeyedVectors
```

```python
import gensim.downloader as api

wv = api.load('word2vec-google-news-300')
```

```python
corpus = df['reviewText']
```

```python
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess

words = []
for sent in corpus:
  sent_token = sent_tokenize(sent)
  for sent in sent_token:
    words.append(simple_preprocess(sent))
```

```python
words
```

```python
import gensim

model = gensim.models.Word2Vec(
  sentences = words,
  vector_size = 300,
)
```

```python
## To get all the vocabulary
model.wv.index_to_key
```

```python
import numpy as np

vocabulary = model.wv.index_to_key

def avg_word2vec(doc):

  # Remove Out-Of-Vocabulary words
  # sent = [word for word in doc if word not in model.wv.index_to_key] # vocabulary = model.wv.index_to_key 
  # print(sent)

  return np.mean([model.wv[word] for word in doc if word in vocabulary], axis=0)
```

```python
from tqdm import tqdm

## apply for the entire sentence

X = []

for i in tqdm(range(len(words))):
  print("Hello", i)
  X.append(avg_word2vec(words[i]))
```
