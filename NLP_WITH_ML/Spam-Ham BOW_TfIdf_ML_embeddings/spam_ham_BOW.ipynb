```python
import pandas as pd
messages = pd.read_csv('spam.csv')
```

```python
messages
```

#### Take ind. feature X = messages as above. Create embeddings using BOW.

```python
import re
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

corpus = []
for i in range(0, len(messages)):
  review = re.sub('[^a-zA-z]', ' ', messages['message'][i]) ## for each sentence take only alphabets
  review = review.lower()
  review = review.split() ## split all words
  
  review = [lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')] ## remove all stopwords

  review = ' '.join(review) ## join back
  corpus.append(review)
```

```python
corpus
```

```python
y = pd.get_dummies(messages['label'])
```

```python
y = y.astype('int')
y
```

```python
## Take the 'ham' row only so we have all containing 0s and 1s for overall data
y = y.iloc[:, 0].values
```

### Take output feature y = label as above. Convert it into 0 and 1 as they are just output so we don't need to embed it.

```python
## Train-Test Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(corpus, y, test_size = 0.20)
```

## Create Bag Of Words for Embeddings

```python
## Create BOG
from sklearn.feature_extraction.text import CountVectorizer
## for Binary BOW enable binary=True
## Here we go with non-binary that means count of words (ex. [2, 1, 0] count is considered here) will be considered
cv = CountVectorizer(
  max_features = 2500, ## this means the top 2500 words having the maximum frequency = max_features
  ngram_range = (1,2)
)
```

### X = independent features
### Embedded using BOW. 

```python
X_train = cv.fit_transform(X_train).toarray()
X_test = cv.transform(X_test).toarray() # to prevent data leakage
```

## Model Training

```python
from sklearn.naive_bayes import MultinomialNB ## Supervised ML
```

```python
spam_detect_model = MultinomialNB().fit(X_train, y_train)
```

```python
spam_detect_model
```

```python
y_pred = spam_detect_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
accuracy_score(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```
