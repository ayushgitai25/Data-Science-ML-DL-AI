```python
import pandas as pd
messages = pd.read_csv('spam.csv')
```

```python
messages
```

```python
import re
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()

corpus = []
for i in range(0, len(messages)):
  review = re.sub('[^a-zA-z]', ' ', messages['message'][i]) ## for each sentence take only alphabets
  review = review.lower()
  review = review.split() ## split all words
  
  review = [lemmatizer.lemmatize(word) for word in review] ## don't apply stopwords

  review = ' '.join(review) ## join back
  corpus.append(review)
```

```python
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

### Get lists of words for every sentence : and it also lowers the case itself

```python
words = []
for sent in corpus:
  sent_token = sent_tokenize(sent)
  for sent in sent_token:
    words.append(simple_preprocess(sent))
```

### So we give our own sentences as words for data and vector size we choose as 100. 

```python
model = gensim.models.Word2Vec(
  sentences = words,
  vector_size = 100,
)
```

### Applying Average Word2Vec : For each sentence we will have a 100 dimension vector.
#### Example : I       am       Ayush  => Avg for this sentence
####         [..]    [..]      [..]  =>     [..]  avg of all [0]
####         [..]    [..]      [..]  =>     [..]  avg of all [1]
####         [..]    [..]      [..]  =>     [..]  avg of all [2]
####         [..]    [..]      [..]  =>     [..]x100 avg of all [99]
####         .........100 dimensions each.}

```python
import numpy as np

vocabulary = model.wv.index_to_key

def avg_word2vec(doc):

  # Remove Out-Of-Vocabulary words
  # sent = [word for word in doc if word not in model.wv.index_to_key] # vocabulary = model.wv.index_to_key 
  # print(sent)

  return np.mean([model.wv[word] for word in doc if word in vocabulary], axis=0)
```

```python
y = messages[list(map(lambda x: len(x)>0, corpus))] ## Just checking which all sentences have len>0 so we don't take the ones which do not have 

y = pd.get_dummies(messages['label']) ## Dependent/Output features
y = y.astype('int')
## Take the 'ham' row only so we have all containing 0s and 1s for overall data
y = y.iloc[:, 0].values
```

```python
import pandas as pd
import numpy as np

# Collect all rows in a list
rows = [X[i].reshape(1, -1) for i in range(len(X))]

# Stack them into a single array
data = np.vstack(rows)

# Create the DataFrame
df = pd.DataFrame(data)
```
