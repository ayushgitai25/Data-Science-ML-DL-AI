```python
import pandas as pd
messages = pd.read_csv('spam.csv')
```

```python
messages.head()
```

```python
import re
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

corpus = []
for i in range(0, len(messages)):
  review = re.sub('[^a-zA-z]', ' ', messages['message'][i]) ## for each sentence take only alphabets
  review = review.lower()
  review = review.split() ## split all words
  
  review = [lemmatizer.lemmatize(word) for word in review if not word in stopwords.words('english')] ## remove all stopwords

  review = ' '.join(review) ## join back
  corpus.append(review)
```

```python
corpus
```

## TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
  max_features = 100,
)
```

```python
X = tfidf.fit_transform(corpus).toarray()
```

```python
for x in X:
  print(x)
```

```python
tfidf = TfidfVectorizer(
  max_features = 100,
  ngram_range = (2,2)
)


X = tfidf.fit_transform(corpus).toarray()

tfidf.vocabulary_
```

```python
for x in X:
  print(x)
```
