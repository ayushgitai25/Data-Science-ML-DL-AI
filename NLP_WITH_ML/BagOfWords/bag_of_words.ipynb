```python
import pandas as pd
messages = pd.read_csv('spam.csv')
```

```python
messages.head()
```

### Data Cleaning and Pre-processing

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
```

```python
corpus = []
for i in range(0, len(messages)):
  review = re.sub('[^a-zA-z]', ' ', messages['message'][i]) ## for each sentence take only alphabets
  review = review.lower()
  review = review.split() ## split all words
  
  review = [ps.stem(word) for word in review if not word in stopwords.words('english')] ## remove all stopwords

  review = ' '.join(review) ## join back
  corpus.append(review)
```

```python
corpus
```

## Create the Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
  max_features = 100,  ## top 100 words having max frequency
  binary = True ## Binary BOW
)
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
X.shape
```

```python
for x in X:
  print(x)
```

### N-gram

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(
  max_features = 100,  ## top 100 words having max frequency
  binary = True, ## Binary BOW

  ngram_range = (1,2)   ## Bi-gram : we will see combination of 2 words in vocabulary
)

X = cv.fit_transform(corpus).toarray()

for x in X:
  print(x)
```

```python
cv.vocabulary_
```
