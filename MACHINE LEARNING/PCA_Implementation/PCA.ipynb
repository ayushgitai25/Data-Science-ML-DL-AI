```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer_dataset = load_breast_cancer()
```

```python
print(cancer_dataset.DESCR)
```

```python
cancer_dataset.keys()
```

```python
df = pd.DataFrame(cancer_dataset['data'], columns=cancer_dataset['feature_names'])
df.head()
```

```python
## Standardization
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()  # Initialize scaler
scaled_data = scaler.fit_transform(df)  # Now transform works
```

```python
pd.DataFrame(scaled_data)
```

```python
## Apply PCA
from sklearn.decomposition import PCA
```

```python
pca = PCA(
  n_components = 2
)
```

```python
data_pca = pca.fit_transform(scaled_data)
```

```python
pd.DataFrame(data_pca)
```

```python
pca.explained_variance_
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
  x = data_pca[:,0],
  y = data_pca[:,1],
  c = cancer_dataset['target'],
  cmap = 'plasma'
)
plt.xlabel('PC - 1') ## max variance
plt.ylabel('PC - 2')
```

```python
import plotly.express as px

# Apply PCA
pca = PCA(n_components=3)
data_pca = pca.fit_transform(scaled_data)

# Create a 3D scatter plot
fig = px.scatter_3d(
    x=data_pca[:, 0],
    y=data_pca[:, 1],
    z=data_pca[:, 2],  # Third principal component
    color=cancer_dataset['target'],  # Color by target variable
    labels={"x": "PC - 1", "y": "PC - 2", "z": "PC - 3"},  # Label axes
    title="3D PCA Visualization"
)

fig.show()
```
