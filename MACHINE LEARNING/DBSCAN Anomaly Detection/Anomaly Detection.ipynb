```python
from sklearn.cluster import DBSCAN
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.datasets import make_circles
```

```python
X, y = make_circles(
  n_samples = 750,
  factor = 0.3,
  noise = 0.1
)
```

```python
pd.DataFrame(X)
```

```python
plt.scatter(X[:, 0], X[:, 1])
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
  eps = 0.10
)
```

```python
X_predicted = dbscan.fit_predict(X)
```

```python
X_predicted
```

```python
pd.DataFrame(X_predicted).value_counts()
```

# We have 52 -1 points as outliers.

### Let's see all the clusters first.

```python
plt.scatter(X[:, 0], X[:, 1], c = X_predicted)
```

### Let's highlight the outliers.

```python
import numpy as np
outliers_index = np.where(X_predicted < 0)
outliers_index
```

## With showing clusters and outliers together.

```python
plt.scatter(X[:, 0], X[:, 1], c = X_predicted)

plt.scatter(X[outliers_index, 0], X[outliers_index, 1], edgecolors = "r")
```

## Without showing clusters and outliers together.

```python
plt.scatter(X[:, 0], X[:, 1])

plt.scatter(X[outliers_index, 0], X[outliers_index, 1], edgecolors = "r")
```
