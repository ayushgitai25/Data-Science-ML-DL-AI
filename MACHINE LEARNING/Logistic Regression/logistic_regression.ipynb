```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
## create the dataset
X, y = make_classification(n_samples = 1000, n_features=10, n_classes=2, random_state=42) # binaty classification so 2 classes
```

```python
X
```

```python
pd.DataFrame(X)
```

```python
# Train-Test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.30, random_state=42)
```

```python
# Model training
from sklearn.linear_model import LogisticRegression
logistic = LogisticRegression()
```

```python
logistic.fit(X_train, y_train)
```

```python
y_pred = logistic.predict(X_test)
print(y_pred)
```

```python
logistic.predict_proba(X_test)
```

```python
logistic.predict_log_proba(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix, classification_report
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
cm = confusion_matrix(y_test, y_pred)
print(cm)
print(classification_report(y_test, y_pred))
```

# Hyper-parameter tuning and Cross Validation

```python
model = LogisticRegression()
penalty = ['l1', 'l2', 'elasticnet']
c_values = [100, 10, 1.0, 0.1, 0.01]
solver = ['newton-cg', 'lbfgs', 'liblinar', 'sag', 'saga']
```

```python
params = dict(penalty=penalty, C=c_values, solver=solver)
```

```python
from sklearn.model_selection import StratifiedKFold    
cv = StratifiedKFold()    # 5 cross validation
```

```python
## GridSearchCV
from sklearn.model_selection import GridSearchCV
grid=GridSearchCV(estimator=model, param_grid=params, scoring='accuracy', cv=cv, n_jobs=-1)
```

```python
grid
```

```python
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred = grid.predict(X_test)
```

```python
score = accuracy_score(y_test, y_pred)
print(score)
cm = confusion_matrix(y_test, y_pred)
print(cm)
print(classification_report(y_test, y_pred))

# older values
# 0.8571428571428571
# [[293  50]
#  [ 50 307]]
#               precision    recall  f1-score   support

#            0       0.85      0.85      0.85       343
#            1       0.86      0.86      0.86       357

#     accuracy                           0.86       700
#    macro avg       0.86      0.86      0.86       700
# weighted avg       0.86      0.86      0.86       700
```
