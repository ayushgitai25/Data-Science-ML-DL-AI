```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import pickle
```

```python
## Load dataset
data = pd.read_csv('churn.csv')
```

```python
data.head()
```

```python
## Pre-processing
data = data.drop(['RowNumber', 'CustomerId', 'Surname'], axis=1)
data
```

```python
data['Gender'].value_counts()
```

```python
data['Geography'].value_counts()
```

## For Binary values we can use Label Encoder.

```python
## Encode categorical variables = Geography and Gender
label_encoder_gender = LabelEncoder()
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
data
```

## For Multi-class values we have to use One-Hot Encoder.

```python
from sklearn.preprocessing import OneHotEncoder

one_hot_encoder = OneHotEncoder(sparse_output = False)

one_hot_encoder_geo = one_hot_encoder.fit_transform(data[['Geography']])

one_hot_encoder_geo
```

```python
one_hot_encoder.get_feature_names_out(['Geography']) ## the same thing will be used while prediction
```

```python
geo_one_hot_encoded_columns = pd.DataFrame(one_hot_encoder_geo, columns=['Geography_France', 'Geography_Germany', 'Geography_Spain'])
```

```python
geo_one_hot_encoded_columns
```


### Operation        |  	axis=0 (Vertical)	         |              axis=1 (Horizontal)
### pd.concat	   |   Stacks rows (adds more rows)	  |    Stacks columns (adds more columns)
### drop       |           	Drops rows	           |                  Drops columns

```python
data = pd.concat([data.drop(['Geography'], axis=1), geo_one_hot_encoded_columns], axis=1)
```

```python
data
```

```python
## Save encoders
with open('label_encoder_gender.pkl', 'wb') as file:
  pickle.dump(label_encoder_gender, file)

with open('one_hot_encoder.pkl', 'wb') as file:
  pickle.dump(one_hot_encoder, file)
```

```python
## divide data into ind. and dep. features
X = data.drop(['Exited'], axis=1)
y = data['Exited']
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(
  X, y, test_size = 0.20, random_state = 42
)
```

### Only perform scaling after seperating the test ans train data so to make sure that train data does not get scaled by test data, scaling whole data will result in data leakage, so first split and then perform scaling seperately on data.

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
pd.DataFrame(X_train) ## 80% of data after scaling
```

```python
with open('scaler.pkl', 'wb') as file:
  pickle.dump(scaler, file)
```

```python
import tensorflow as tf

from tensorflow.keras.models import Sequential  # Sequential Model
from tensorflow.keras.layers import Dense       # Create a neuron or node
```

```python
number_of_input_nodes = X_train.shape[1]  ## 12
```

```python
## Build our ANN model
model = Sequential([
       Dense(64, activation= 'relu', input_shape= (number_of_input_nodes, )), ## HL 1 connected with I/P layer with 64 nodes and input of 12 nodes
       Dense(32, activation= 'relu'), ## HL 2 with 32 nodes
       Dense(1, activation = 'sigmoid') ## O/P layer with 1 node
  ]
)
```

### Number of parameters :
### 12*64+64 = 832
### 64*32+32 = 2080
### 32*1+1 = 33

```python
model.summary()
```

```python
opt = tf.keras.optimizers.Adam(learning_rate = 0.01)
loss = tf.keras.losses.BinaryCrossentropy()
```

```python
## compile the model
model.compile(
  optimizer = opt,
  loss = loss, ## binary classification
  metrics = ['accuracy'] ## the accuracy we usually use in Classification problems
)
```

```python
## Set up the Tensorboard

from tensorflow.keras.callbacks import EarlyStopping, TensorBoard ## to visualize all the logs
import datetime

log_dir = "logs/fit/" + datetime.datetime.now().strftime("%Y%m%d-%H%M%S")
tf_callback = TensorBoard(log_dir= log_dir, histogram_freq=1)
```

### If we are training and after certain epochs we see loss is not decreasing any more, we can use EarlyStopping.

```python
## Set up Early Stopping 
early_stopping_callback = EarlyStopping(
  monitor='val_loss', ## validation loss
  patience=10, ## wait for 10 epochs
  restore_best_weights=True)
```

```python
## Training the model
history = model.fit(
  X_train, y_train,
  validation_data = (X_test, y_test),
  epochs=100,
  callbacks=[tf_callback, early_stopping_callback]
)
```

```python
model.save('model.h5')
```
