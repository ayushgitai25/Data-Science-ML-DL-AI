```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences = [
    "The morning sun warmed the quiet street",
    "He placed the keys gently on the table",
    "The library was silent except for turning pages",
    "She tied her hair back before starting to paint",
    "The smell of fresh bread filled the kitchen",
    "Rainwater trickled down the old brick wall",
    "They watched the fireworks light up the sky",
    "The waves left seashells scattered along the shore",
    "A cat stretched lazily in the afternoon sun",
    "The train whistle echoed through the valley"
]
```

```python
## define the vocab_size
vocab_size = 10000
```

```python
### One hot representation
one_hot_repr = [one_hot(words, vocab_size) for words in sentences]
one_hot_repr
```

## What does this mean?
### If vocab_size = 10000,
### for first word of sentence 1 : we have a vector [0, 0, 0, ..., 0, 1, 0, ..., 0]   # dimension = 10000
### The 1 is at position 6275. All other positions are 0.

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
import tensorflow as tf
```

```python
## Make all sentences of same length as RNN will take fixed number of timestamps
sent_len = 9
embedded_docs = pad_sequences(
  one_hot_repr,
  padding='pre', ## Add 0s at start
  maxlen=sent_len
)

embedded_docs ## All sentences of len=9
```

```python
## feature representation
dim=10
```

```python
model = Sequential()
model.add(Embedding(vocab_size, dim, input_length = sent_len)) ## internally it uses Word2Vec for embedding creation
model.compile(
  optimizer = tf.keras.optimizers.Adam(learning_rate=0.01),
  loss = tf.keras.losses.MSE
)
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```

```python
model.predict(embedded_docs).shape
```
