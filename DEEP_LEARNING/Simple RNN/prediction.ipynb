```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
model = load_model('simple_rnn_imdb.h5')
model.summary()
```

```python
model.get_weights()
```

```python
print("WITHOUT BIAS ONLY WEIGHTS:")
print(f"Embedding layer weights: {len(model.get_weights()[0])*len(model.get_weights()[1])}")
print(f"Input Layer weights which is passing one word at a time of dimension: {len(model.get_weights()[2])}")
print(f"Simple RNN Hidden Layer weights: {len(model.get_weights()[2])*len(model.get_weights()[3]) + len(model.get_weights()[3])*len(model.get_weights()[4])}")
print(f"Output Layer weights: {len(model.get_weights()[4]) * len(model.get_weights()[5])}")
```

```python
## Just for understanding map index back to words
word_index = imdb.get_word_index()
word_index
```

```python
## Function to preprocess user input
def preprocess_text(text):
  words = text.lower().split()
  encoded_review = [word_index.get(word, 2) + 3 for word in words]
  padded_review = sequence.pad_sequences([encoded_review], maxlen=500)
  return padded_review
```

```python
## Predict function
def predict_sentiment(review):
  preprocessd_input = preprocess_text(review)

  prediction = model.predict(preprocessd_input)

  sentiment = "Positive" if prediction[0][0] > 0.5 else "Negative" ## As Output Layer has sigmoid so threshold = 0.5

  return sentiment, prediction[0][0]
```

```python
example_review = "This move was fantastic! The acting was great and the plot was thrilling."

sentiment, score = predict_sentiment(example_review)

print(f"Review : {example_review}")
print(f"Sentiment:  {sentiment}")
print(f"Prediciton Score : {score}")
```
